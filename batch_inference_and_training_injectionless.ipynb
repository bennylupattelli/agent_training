## 260929 — gamma isolated

Same schedules, `max_steps` and seeds as 260914; `lambd` and `time_horizon` reverted to the 260723
values. The only difference from a 260723-style run is extrinsic gamma at 0.998.

If this reproduces 260914's result (one central cluster, `p_high` near 1, few degenerate runs) then
gamma alone is enough and `lambd` and `time_horizon` need not be credited. If it looks like 260723
instead, the matched credit-assignment package mattered and the next arms should add `lambd` and
`time_horizon` back one at a time.

Seeds come from `DEFAULT_TRAINING_SEEDS`, the seeds 260914 was trained on, so model 0007 here is
the same seed as `260914_0007` and the two batches compare pairwise. Leave `seeds` unset.

At 7e6 steps each run takes hours, so treat the model count as the compute decision. Twenty gives a
usable read on the fraction of degenerate runs, which is the measure gamma moved most (44% of 260723
runs were peripheral against 3% of 260914). Ten is enough to see a large effect and not much else.
See `READMEs/training_hyperparameters.md`.

```python
from pathlib import Path

from trainer_and_simulator_functions import sequential_runs

sequential_runs(
        in_yaml=Path("~/Unity/Octagon/Assets/Scripts/MLConfigFiles/SocialConfig260929.yaml").expanduser(),
        run_dir=Path("~/Unity/Octagon").expanduser(),
        gamma=0.998,                 # the variable under test; 260723 used 0.99
        sp=1e-3,
        behaviour_name="OctagonAgentSocial",
        # the 260914 build, held fixed across arms so the only difference is the config
        unity_build=Path("/home/tom/Unity/Octagon/builds/260914/build_261014_01_social.x86_64"),
        base_run_id="261020",
        device="cuda:0",
        n_models=50,
        run_id_offset=0,  # models already finished; trains from model offset+1. To resume model N, use N-1
        simulate=False,
        n_envs=16,
        # extra_args=["--resume"] # edit to use this only on the first model of batch

        # seeds defaults to DEFAULT_TRAINING_SEEDS, so this batch is paired with 260914.
        # Only pass seeds=None if a batch is deliberately not meant to be compared pairwise.
        # 100 seeds are on the list, so run_id_offset + n_models must not exceed 100.
        #
        # No extra_args. 260929_0001 is a fresh run id, and mlagents-learn --resume on a run id
        # with no results directory raises "Previous data from this run ID was not found", which
        # would stop the batch on its first model. Add extra_args=["--resume"] only when the first
        # model of the batch is being continued.
)
```
