## 261020 — opponent sensing (261014 build), gamma 0.99

`SocialConfig261020.yaml` is the 260929 config with extrinsic gamma back at 0.99, the 260723 value.
`lambd` 0.95 and `time_horizon` 128 already match gamma 0.99. Schedules (constant learning rate and
epsilon, linear beta) and `step_penalty` 1e-3 are held at the 260929 values. `max_steps` is 4e6, the
260723 value, as a gamma 0.99 run levels off by about 3e6. Beta decays to zero by 4e6 rather than 7e6. The
build is `builds/261014/build_261014_social_01`.

In the 261014 build both seats sense each other, walls hide the opponent, and the opponent value is
proximity (1 - d/60) rather than distance. Before it, the `OpponentAgent` seat had no opponent input,

batch asks whether full access during training leads the policy to use it. See
`READMEs/opponent_sensing.md`.

peripheral runs, against 6 of 33 for 260929 (gamma 0.998). The earlier opponent-response test used
peripheral models, so this batch can repeat it with both seats sensing.

No earlier batch matches this one in everything but the build. 260723 differs in the learning rate
and epsilon schedules and the build, and 260929 differs in gamma, `max_steps` and the build. Whether the networks use the opponent
input can be tested within this batch, and the earlier batches serve as a reference.

Inference and tournaments for these models need the 261014 build. Keep older models in their own
builds, as they read the opponent channel the other way round.

Seeds come from `DEFAULT_TRAINING_SEEDS`, so model 0007 here trains on the same seed as
`260929_0007` and `260914_0007`. Leave `seeds` unset.

## 261029 — solo split-cost pilot (261029 build), gamma 0.99

First batch with separate action costs. `step_penalty` stays at 1e-3 in every cell (the 260723 /
261020 time cost). `translation_cost` and `rotation_cost` are added on top, each varied alone at 1e-4,
3e-4, 1e-3 and 3e-3, plus a (0, 0) baseline. That's 9 cells × 3 seeds = 27 runs. Hyperparameters are
the 261020 set with self-play removed, and schedules are constant. The aim is the distribution of
positional strategies (onset centrality) per cell. See `READMEs/split_action_costs.md` for why these
values were chosen.

- Configs: `configs/split_cost_261029/SoloSplitCost261029_<cell>.yaml`, generated from
  `Octagon/Assets/Scripts/MLConfigFiles/SoloSplitCostConfig261027.yaml` (only the costs differ).
- Build: `builds/261029/build_261029_solo_01`, the first build with the cost code. Older builds
  ignore the new keys and older configs throw on this build.
- Run ids: `261029_<cell>_000N`. Every cell trains models 1-3 on the same three
  `DEFAULT_TRAINING_SEEDS` (also the seeds of `261020_0001`-`0003`), so cells pair by seed.
- Order: the extremes go first, so the direction of each effect shows up early.
- Time: about 2 h per run (260515 solo took 1.56 h for 3e6 steps), so about 2-2.5 days in all.
- Inference needs the matching cell config, since all three costs must be present in
  `environment_parameters`.

To restart part-way, trim `CELLS` to the cells not yet started, and for the cell in progress set
`run_id_offset` to the number of finished models (add `extra_args=["--resume"]` to continue a
partly trained model).

```python
from pathlib import Path

from trainer_and_simulator_functions import sequential_runs

SPLIT_COST_CONFIGS = Path("~/repos/agent_training/configs/split_cost_261029").expanduser()

# extremes first, so the direction of each effect is visible before the batch finishes
CELLS = ["base", "T3e-3", "R3e-3", "T1e-3", "R1e-3", "T3e-4", "R3e-4", "T1e-4", "R1e-4"]

for cell in CELLS:
    sequential_runs(
            in_yaml=SPLIT_COST_CONFIGS / f"SoloSplitCost261029_{cell}.yaml",
            run_dir=Path("~/Unity/Octagon").expanduser(),
            gamma=0.99,                  # printed only; the yaml sets gamma and all costs (yaml_injection is False)
            sp=1e-3,                     # printed only
            behaviour_name="OctagonAgentSolo",
            # the 261029 build, the first with translation_cost / rotation_cost
            unity_build=Path("/home/tom/Unity/Octagon/builds/261029/build_261029_solo_01.x86_64"),
            base_run_id=f"261029_{cell}",
            device="cuda:0",
            n_models=3,
            run_id_offset=0,  # models already finished; trains from model offset+1. To resume model N, use N-1
            simulate=False,
            n_envs=16,
            # seeds defaults to DEFAULT_TRAINING_SEEDS: models 1-3 of every cell share seeds
            # extra_args=["--resume"] # only when continuing the first model of a cell
    )
```

### 261029 results (characterised 2026-11-03)

All 27 runs finished at 4e6 steps (2026-10-31) and are filed in `Octagon/results/2610/261029/`.
Inference logs are in `Octagon/simulations/2610/261029/`. Full tables are in
`READMEs/split_action_costs.md` (section 8) and `results/2610/261029/notes.txt`.

**Inference conditions:** 201 episodes per model with the final checkpoint,
`--trial_seq trial_sequences/trials_100000_seed17.json`, `--seed 17`, the 261029 solo build, and each
model's own cell config. It was run from the shell, not this notebook, because `batch_inference` cannot
pass `--trial_seq` at present. It was characterised with `octagon_analysis/analysis/model_characterisation.py`
(`players=(0,)`).

| cell | central / peripheral | onset centrality | chose Low | median onset to trigger |
| --- | --- | --- | --- | --- |
| (0, 0) | 3 / 0 | 0.90-0.91 | 0-10% | 1.9-2.1 s |
| T 1e-4 to 1e-3 | 9 / 0 | 0.90-0.96 | 0-17% | 2.0 s |
| T 3e-3 seed 1 (collapsed, 1 of 199 trials) | 0 / 1 | 0.07 | 100% | 7.2 s |
| T 3e-3 seed 2 (works, 196 of 199 trials) | 0 / 1 | 0.05 | 32% | 2.8 s |
| T 3e-3 seed 3 (collapsed, 26 of 199 trials) | 0 / 1 | 0.05 | 31% | 21 s |
| R 1e-4, 3e-4 | 6 / 0 | 0.79-0.92 | 0-23% | 2.0-2.2 s |
| R 1e-3 | 2 / 1 | 0.84, 0.03, 0.79 | 20-26% | 1.9-3.2 s |
| R 3e-3 | 0 / 3 | 0.02-0.04 | 17-35% | 3.0-3.8 s |

**Simple interpretations:**
- **The solo baseline is central.** The ~50/50 split came from social self-play.
- **A high rotation cost makes agents peripheral, as hypothesised.** `R3e-3` agents wait at the back
  of an alcove facing the centre, and almost stop turning in the ITI (1-3% of baseline). This costs
  them longer routes and more Low choices. Their longer episodes in TensorBoard (461-502 steps) were
  the first sign.
- **The switch to peripheral lies between rotation 3e-4 and 3e-3.** At 1e-3 a turning step costs as
  much as a waiting step, and one of the three seeds is already peripheral.
- **Translation cost keeps agents central up to 1e-3.** They save by resting in the centre during the
  ITI rather than by moving to the wall.
- **At `T3e-3` two seeds collapsed and one camps at the wall.** Seeds 1 and 3 rest until the timeout
  on most trials. Seed 2 works (196 of 199 trials, High chosen on 68%, 2.8 s). It never moves between
  trials, and waits where it last triggered, facing roughly into the arena. This is the predicted wall
  camper, and the only working peripheral strategy that translation cost alone produced. With one seed
  it is unknown whether camping is reliable somewhere between 1e-3 and 3e-3. Most rotation can be
  avoided, but the trip to a wall cannot. At 3e-3 the trip costs more than a discounted Low reward is
  worth.
- These are 3 seeds per cell, so treat the shares as rough.

**Next steps:**
1. Sample the critical ranges more finely: rotation between 3e-4 and 3e-3 with more seeds, and
   translation at ~2e-3 to see whether the wall camping is reliable below the cost where learning
   collapses. 3e-3 is too high to use as a level.
2. A combined grid with both costs non-zero (not done yet).
3. The same costs in social training.


## 261104 — solo split-cost seed extension (261029 build), gamma 0.99

More seeds for the three cells where the 261029 pilot found peripheral strategies, before moving on
to a combined grid. Each cell gets 10 new seeds.

- `T2e-3` (translation 2e-3, rotation 0) replaces `T3e-3`, where 2 of 3 seeds collapsed and one camped
  at the last wall it triggered. It tests whether camping is reliable below the collapse cost.
- `R1e-3` (1 of 3 peripheral in 261029) and `R3e-3` (3 of 3) are unchanged costs with more seeds.
  They test whether the rotation agents really cross to the far side to face inward without turning.

Settings:
- Configs: `configs/split_cost_261104/SoloSplitCost261104_<cell>.yaml`. The R configs match 261029
  apart from the header.
- Build: `builds/261029/build_261029_solo_01`, the same as 261029.
- Run ids are `261104_<cell>_0004` to `_0013`, on `DEFAULT_TRAINING_SEEDS` positions 4-13, which come
  after the three the pilot used. Model N of every cell has the same seed, so `R1e-3` and `R3e-3` pool
  with 261029 models 0001-0003 to give 13 seeds each. `T2e-3` has no models 0001-0003.
- Order: interleaved by seed. Each seed trains all three cells before the next seed starts, so a
  batch stopped early still has the same seeds in every cell.
- Time: about 2 h per run, so 30 runs take about 2.5 days.
- Notes: `Octagon/results/2611/261104/notes.txt`.

To restart part-way, set `FIRST_MODEL` to the first model number not finished in every cell, and train
any cell missing that model on its own first (add `extra_args=["--resume"]` to continue a partly
trained model).


```python
from pathlib import Path

from trainer_and_simulator_functions import sequential_runs

SPLIT_COST_CONFIGS = Path("~/repos/agent_training/configs/split_cost_261104").expanduser()

CELLS = ["T2e-3", "R1e-3", "R3e-3"]
FIRST_MODEL, LAST_MODEL = 4, 13     # DEFAULT_TRAINING_SEEDS positions 4-13; 1-3 were used by 261029

# interleave by seed, so a batch stopped early has the same seeds in every cell
for model in range(FIRST_MODEL, LAST_MODEL + 1):
    for cell in CELLS:
        sequential_runs(
                in_yaml=SPLIT_COST_CONFIGS / f"SoloSplitCost261104_{cell}.yaml",
                run_dir=Path("~/Unity/Octagon").expanduser(),
                gamma=0.99,                  # printed only; the yaml sets gamma and all costs (yaml_injection is False)
                sp=1e-3,                     # printed only
                behaviour_name="OctagonAgentSolo",
                unity_build=Path("/home/tom/Unity/Octagon/builds/261029/build_261029_solo_01.x86_64"),
                base_run_id=f"261104_{cell}",
                device="cuda:0",
                n_models=1,
                run_id_offset=model - 1,     # trains model `model` on DEFAULT_TRAINING_SEEDS[model - 1]
                simulate=False,
                n_envs=16,
                # extra_args=["--resume"] # only when continuing a partly trained model
        )
```

## 261109 — solo combined split-cost grid (261029 build), gamma 0.99

First grid with both action costs non-zero, after the 261029 pilot and 261104 seed extension found
where each cost alone moves agents (README sections 8-9).

| | R 3e-4 | R 1e-3 | R 3e-3 |
| --- | --- | --- | --- |
| **T 1e-4** | T1e-4_R3e-4 | T1e-4_R1e-3 | T1e-4_R3e-3 |
| **T 3e-4** | T3e-4_R3e-4 | T3e-4_R1e-3 | T3e-4_R3e-3 |
| **T 1e-3** | T1e-3_R3e-4 | T1e-3_R1e-3 | T1e-3_R3e-3 |

- **Main question:** does a translation cost turn rotation-cost crossers into campers? Crossing an
  ITI is ~125 translating steps; turning 180 deg and camping is 25 rotating steps. Crossing should
  stop paying at translation ~6e-4 under R3e-3 and ~2e-4 under R1e-3, both inside the grid.
- **R 3e-4** is sub-threshold alone (0/3 peripheral). It tests whether the costs add up.
- **Translation is capped at 1e-3.** The trip to a wall cannot be avoided, and with a rotation cost on
  top a Low choice is near zero value at 2e-3 (collapse risk). Rotation goes to 3e-3, as most turning
  can be avoided.
- The single-cost cells of 261029/261104 are the grid's zero row and column, so they are not retrained.

Settings:
- Configs: `configs/split_cost_261109/SoloSplitCost261109_<cell>.yaml`. Hyperparameters match 261104;
  only the costs differ (`step_penalty` 1e-3 in every cell).
- Build: `builds/261029/build_261029_solo_01`, the same as 261029 and 261104.
- Run ids `261109_<cell>_0001` to `_0005`, on `DEFAULT_TRAINING_SEEDS` positions 1-5. These pair with
  the edge cells: seeds 1-3 for every single-cost cell, 1-5 for R1e-3 and R3e-3.
- Order: interleaved by seed, so a batch stopped early has the same seeds in every cell. Within a
  seed the R3e-3 and R1e-3 columns go first, as they answer the main question.
- Time: about 2 h per run, so 45 runs take about 3.75 days.
- Notes: `Octagon/results/2611/261109/notes.txt`.

To restart part-way, set `FIRST_MODEL` to the first model number not finished in every cell, and train
any cell missing that model on its own first (add `extra_args=["--resume"]` to continue a partly
trained model).
